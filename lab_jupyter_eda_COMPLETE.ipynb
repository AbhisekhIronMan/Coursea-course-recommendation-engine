### TASK: Analyze Course Genres

_TODO: Find all courses with genre `MachineLearning` == 1_

```python
ml_courses = course_df[course_df['MachineLearning'] == 1]
ml_courses
```

_TODO: Find all courses with genres `MachineLearning` == 1 and `BigData` == 1_

```python
ml_bd_courses = course_df[(course_df['MachineLearning'] == 1) & (course_df['BigData'] == 1)]
ml_bd_courses
```

```python
genres = course_df.columns[2:]
genres
```

_TODO: Use the Pandas dataframe sum() and sort_values() methods to generate a sorted course count per genre._

```python
genre_sums = course_df[genres].sum(axis=0)
genre_sums = pd.DataFrame(genre_sums, columns=['Count'])
genre_sums = genre_sums.sort_values(by="Count", ascending=False)
genre_sums
```

_TODO: Use seaborn barplot to plot course genre counts using a barchart._

```python
plt.figure(figsize=(12, 6))
plot = sns.barplot(x=genre_sums.index, y="Count", data=genre_sums)
plot.set_xticklabels(plot.get_xticklabels(), rotation=90)
plt.title("Course Counts per Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Courses")
plt.tight_layout()
plt.show()
```

### Task: Find the Top-20 Most Popular Courses

_TODO: Use Pandas groupby() and size() to get top 20 courses by enrollment count._

```python
course_rating_counts = ratings_df.groupby(['item']).size().reset_index()
course_rating_counts.columns = ['course', 'Ratings']
course_rating_counts = course_rating_counts.sort_values(by='Ratings', ascending=False).reset_index(drop=True)
top_courses = course_rating_counts[:20]
top_courses
```

_TODO: Use Pandas merge() method to join the course_df with course titles._

```python
top_courses = pd.merge(top_courses, course_df[['COURSE_ID', 'TITLE']], how='left', left_on='course', right_on='COURSE_ID')
top_courses = top_courses[['TITLE', 'Ratings']]
top_courses
```

```python
total = ratings_df.shape[0]
total
```

_TODO: Get the percentage of the top-20 course enrollments._

```python
top = top_courses['Ratings'].values.sum()
percentage = (top * 100) / total
print(f"Percentage of the top course enrollments {round(percentage, 2)}%")
```
